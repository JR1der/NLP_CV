```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import guidance
import torch

model_name = "deepseek-ai/DeepSeek-R1-Distill-Qwen-7B"

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.float16,
    device_map="auto",
    trust_remote_code=True
)
tokenizer = AutoTokenizer.from_pretrained(model_name)

llm = guidance.llms.Transformers(model, tokenizer)
```

```python
from pdf_extractor import extract_text_from_pdf

filename = "CV_IvanPopovych.pdf"

cv_text = extract_text_from_pdf(filename)

cv_prompt = f"""
    You are an AI that extracts skills from people's resume's.
    Extract a list of skills from the following text:
    
    Text: {cv_text}

    Skills:
    """

cv_skills = llm + cv_prompt

cv_skills
```

```python
job_desc = """
    Looking for a frontend developer, needs to know React, TypeScript, MongoDB, Node.js
"""

job_prompt = f"""
You are an AI that extracts skills from job descriptions.
Extract a list of skills from the following text:

Text: {job_desc}

Skills:
"""

job_skills = llm + job_prompt

job_skills
```

```python
match_prompt = f"""
    You are an AI that evaluates a CV against a job description.

    CV: {cv_text}

    Job Description: {job_desc}

    Provide a similarity score (0-100%) and list missing skills:
"""

match_result = llm + match_prompt

match_result
```
